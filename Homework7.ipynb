W7A16:
Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
from collections import Counter

def make_beautiful_array(arr):
    counts = Counter(arr)
    y = 0
    for x, f in counts.items():
        if f < x:
            y += f  # Xóa toàn bộ vì không đủ x lần
        else:
            y += (f - x)  # Xóa phần dư thừa
    return y
```

W7A17:
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. 


```python
def find_majority_element(nums):
    candidate = None
    count = 0
    # Giai đoạn 1: Tìm ứng cử viên
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)
    
    # Giai đoạn 2: Kiểm tra lại (vì đề bài nói "nếu có")
    if nums.count(candidate) > len(nums) // 2:
        return candidate
    return None
```

W7A18:
Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng dần, lẻ giảm dần; ;

```python
def conditional_sort(arr):
    evens = sorted([x for x in arr if x % 2 == 0])
    odds = sorted([x for x in arr if x % 2 != 0], reverse=True)
    return evens + odds
```

W7A19:
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không dùng bất cứ thuật toán sắp xếp nào

```python
def merge_arrays(arr1, arr2):
    # Đảm bảo cả 2 mảng đều ở dạng tăng dần trước khi trộn
    a = arr1 if arr1[0] <= arr1[-1] else arr1[::-1]
    b = arr2 if arr2[0] <= arr2[-1] else arr2[::-1]
    
    res = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            res.append(a[i]); i += 1
        else:
            res.append(b[j]); j += 1
    return res + a[i:] + b[j:]
```

W7A22:
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất


```python
def most_frequent_name(names):
    counts = Counter(names)
    # Sắp xếp theo: 1. Tần suất giảm dần (-x[1]), 2. Tên tăng dần (x[0])
    winner = sorted(counts.items(), key=lambda x: (-x[1], x[0]))
    return winner[0][0]
```

W7A23:
Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài >=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí tự cái.



```python
import re

def count_names(text):
    # Regex: Chữ hoa đứng đầu, theo sau là ít nhất 1 chữ thường
    pattern = r'\b[A-Z][a-z]+\b'
    return len(re.findall(pattern, text))
```

W7A24: Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên

```python
def remove_adjacent_duplicates(s):
    words = s.split()
    if not words: return ""
    res = [words[0]]
    for i in range(1, len(words)):
        if words[i] != words[i-1]:
            res.append(words[i])
    return " ".join(res)
```

W7A25:
Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số

```python
def largest_k_digit_number(number_str, k):
    stack = []
    to_remove = len(number_str) - k
    for digit in number_str:
        while to_remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            to_remove -= 1
        stack.append(digit)
    return "".join(stack[:k])
```

W7A26:
Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các số đó


```python
def stats_from_text(text):
    nums = [float(s) for s in re.findall(r'\d+\.?\d*', text)]
    if not nums: return 0, 0
    total = sum(nums)
    return total, total / len(nums)
```

W7A27:
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ hay không

```python
def is_valid_parentheses(s):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}
    for char in s:
        if char in pairs.values():
            stack.append(char)
        elif char in pairs:
            if not stack or stack.pop() != pairs[char]:
                return False
    return not stack
```

W7A28:
Bài toán: Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)

```python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

def longest_composite_sequence(N):
    # Bắt đầu tìm từ N+1
    max_seq = []
    current_seq = []
    # Giới hạn tìm kiếm thực tế tùy thuộc vào N, ở đây minh họa logic
    for i in range(N + 1, N + 1000): 
        if not is_prime(i):
            current_seq.append(i)
        else:
            if len(current_seq) > len(max_seq):
                max_seq = current_seq
            current_seq = []
    return max_seq
```

W7A29:
Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
def longest_continuous_increasing(arr):
    if not arr: return []
    max_seq = []
    current_seq = [arr[0]]
    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            current_seq.append(arr[i])
        else:
            if len(current_seq) > len(max_seq):
                max_seq = current_seq
            current_seq = [arr[i]]
    return max_seq if len(max_seq) >= len(current_seq) else current_seq
```

W7A30:
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập vào từ bàn phím 


```python
import bisect

def longest_increasing_subsequence(arr):
    if not arr: return 0
    tails = []
    for x in arr:
        idx = bisect.bisect_left(tails, x)
        if idx < len(tails):
            tails[idx] = x
        else:
            tails.append(x)
    return len(tails)
```
